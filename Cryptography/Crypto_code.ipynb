```python
class Alphabet():

    morse = {
    'a': '.-', 'b': '-...', 'c': '-.-.', 'd': '-..', 'e': '.',
    'f': '..-.', 'g': '--.', 'h': '....', 'i': '..', 'j': '.---',
    'k': '-.-', 'l': '.-..', 'm': '--', 'n': '-.', 'o': '---',
    'p': '.--.', 'q': '--.-', 'r': '.-.', 's': '...', 't': '-',
    'u': '..-', 'v': '...-', 'w': '.--', 'x': '-..-', 'y': '-.--',
    'z': '--..', ' ': ' '
    }

    morse_bin = {
    'a': '01', 'b': '1000', 'c': '1010', 'd': '100', 'e': '0',
    'f': '0010', 'g': '110', 'h': '0000', 'i': '00', 'j': '0111',
    'k': '101', 'l': '0100', 'm': '11', 'n': '10', 'o': '111',
    'p': '0110', 'q': '1101', 'r': '010', 's': '000', 't': '1',
    'u': '001', 'v': '0001', 'w': '011', 'x': '1001', 'y': '1011',
    'z': '1100', ' ': ' '
    }

    alphabet_dict = {
    'a':1, 'b':2, 'c':3, 'd':4, 'e':5,
    'f':6, 'g':7, 'h':8, 'i':9, 'j': 10,
    'k':11, 'l':12, 'm':13, 'n':14, 'o':15,
    'p':16, 'q':17, 'r':18, 's':19, 't':20,
    'u':21, 'v':22, 'w':23, 'x':24, 'y':25,
    'z':26
    }

    alphabet_str = 'abcdefghijklmnopqrstuvwxyz'

    novoalfabeto = r"&/*}+.$%=(;:,>_-){|[^~<]\`"
    
    def induced_alphabet(sample, spaces = True):
        temp = []

        if spaces == False:
            sample = sample.replace(' ', '')

        for _ in set(sample):
            temp.append(
                max(sample, key = sample.count)
                )
                
            sample = sample.replace(temp[-1:][0], '')

        return ''.join(temp)
```

```python
class Transposition(Criptography):

    def embaralhador(frase):
        letras = list(frase.replace(' ', ''))
        rd.shuffle(letras)
        nova_frase = ''

        for i in frase:
            if i != ' ':
                nova_frase += letras.pop()

            else:
                nova_frase += i

        return nova_frase
    
    def rail_fence(frase):
        frase_final = frase.replace(' ', '')
        bloco_um = ''
        bloco_dois = ''

        for i in range(len(frase_final)):
            if i & 1 == 0:
                bloco_um += frase_final[i]

            else:
                bloco_dois += frase_final[i]

        return bloco_um + ' ' + bloco_dois
    
    def rail_fence_gen(frase, quantas_linhas):
        letras = frase.replace(' ', '')
        linhas = ['' for _ in range(quantas_linhas)]

        for i in range(len(letras)):
            linhas[i%len(linhas)] += letras[i]

        return ' '.join(linhas)
```

```python
print(Transposition.embaralhador('Ola eu estou programando'))
print(Transposition.rail_fence('Ola eu estou programando'))
print(Transposition.rail_fence_gen('Ola eu estou programando', 5))
```

```python
class Substitution(Criptography):

    def to_morse(frase):
        frase_final = ''

        for i in frase:
            if i == ' ':
                frase_final += ' '

            else:
                frase_final += Alphabet.morse[i] + ' '

        return frase_final

    def letras_por_endereco(frase):
        dicionario = str.maketrans({
            Alphabet.alphabet_str[i]: f'{i+1} ' for i in range(len(Alphabet.alphabet_str))
        })


        return frase.translate(dicionario)

    def substituidor(frase):
        substituidor = str.maketrans({
            Alphabet.alphabet_str[x]: Alphabet.novoalfabeto[x] for x in range(len(Alphabet.novoalfabeto))
        })

        return frase.translate(substituidor)

    def palavras_por_coordenadas(frase):
        frase_final = ''

        for i in frase:
            if i == ' ':
                frase_final += ' '

            else:
                frase_final += f'({Alphabet.alphabet_str.index(i)//5 + 1},{Alphabet.alphabet_str.index(i)%5 + 1})'

        return frase_final
```

```python
print(Substitution.to_morse('ola eu estou programando'))
print(Substitution.letras_por_endereco('ola eu estou programando'))
print(Substitution.substituidor('ola eu estou programando'))
print(Substitution.palavras_por_coordenadas('ola eu estou programando'))
```
