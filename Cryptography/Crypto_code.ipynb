```python
class Transposition(Criptography):

    def embaralhador(frase):
        letras = list(frase.replace(' ', ''))
        rd.shuffle(letras)
        nova_frase = ''

        for i in frase:
            if i != ' ':
                nova_frase += letras.pop()

            else:
                nova_frase += i

        return nova_frase
    
    def rail_fence(frase):
        frase_final = frase.replace(' ', '')
        bloco_um = ''
        bloco_dois = ''

        for i in range(len(frase_final)):
            if i & 1 == 0:
                bloco_um += frase_final[i]

            else:
                bloco_dois += frase_final[i]

        return bloco_um + ' ' + bloco_dois
    
    def rail_fence_gen(frase, quantas_linhas):
        letras = frase.replace(' ', '')
        linhas = ['' for _ in range(quantas_linhas)]

        for i in range(len(letras)):
            linhas[i%len(linhas)] += letras[i]

        return ' '.join(linhas)
```
